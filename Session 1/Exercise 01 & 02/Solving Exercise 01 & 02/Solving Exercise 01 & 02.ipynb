    <h4>Write the program that prompts the user for three difrent colors(as strings). if all three colors are the same, print "All colors match". if only two colors match, print"Two colors match", if no colors match print "No colors match".
    <br> (Question 04)

```python
col1 = input("enter the color : ")
col2 = input("enter the color : ")
col3 = input("enter the color : ")

if col1 == col2 == col3 : 
    print("All colors match")
elif col1 == col2 or col2 == col3 or col1 == col3 :
    print("Two colors match") 
else :
    print("No colors match")
```
